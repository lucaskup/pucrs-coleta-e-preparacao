## Parte 3: roteiro de EDA (20 min)

O mediador atribui **uma etapa** a cada grupo. Todos rodam a célula de preparação; depois, cada
grupo aprofunda a sua etapa e prepara **um achado** (um número ou uma tabela) para mostrar na
apresentação, explicando *o que encontrou* e *que decisão isso pede* na preparação dos dados.

| Etapa | Perguntas-guia |
| --- | --- |
| **1. Estrutura** | Quantos objetos e atributos? Algum identificador? Tipo de armazenamento × nível de medição? |
| **2. Qualidade** | Quantos ausentes por coluna (incluindo sentinelas)? Duplicatas? Categorias inconsistentes? O condomínio falta ao acaso? (compare o ano de construção de quem tem e de quem não tem) |
